```python
# ─── Install all Python dependencies ─────────────────────────────────────────
import subprocess, sys

packages = ['streamlit', 'plotly', 'pandas', 'numpy', 'Pillow', 'pytesseract', 'opencv-python', 'PyMuPDF', 'spacy']

print('⚡ Installing GT-Invoice Extraction dependencies...')
for pkg in packages:
    result = subprocess.run([sys.executable, '-m', 'pip', 'install', pkg, '-q'], capture_output=True, text=True)
    print(f'  {"✅" if result.returncode == 0 else "⚠️"} {pkg}')

subprocess.run([sys.executable, '-m', 'spacy', 'download', 'en_core_web_sm', '-q'])
print('✔ GT-Invoice Extraction setup complete!')
```

```python
# ─── GT Engine Core Class Definitions & Windows Tesseract Auto-Detect ─────────
import os, re, json, random, datetime, io
import warnings
warnings.filterwarnings('ignore')

try:
    import pytesseract
    tess_paths = [
        r'C:\Program Files\Tesseract-OCR\tesseract.exe',
        r'C:\Program Files (x86)\Tesseract-OCR\tesseract.exe',
        os.path.expanduser(r'~\AppData\Local\Programs\Tesseract-OCR\tesseract.exe'),
    ]
    for p in tess_paths:
        if os.path.exists(p):
            pytesseract.pytesseract.tesseract_cmd = p
            break
    HAS_TESS = True
except Exception:
    HAS_TESS = False

print(f'⚡ GT Engine Loaded! Tesseract Status: {"✅ Active" if HAS_TESS else "⚠️ Fallback Mode"}')
```

```python
# ─── Launch GT-Invoice Extraction Dashboard ──────────────────────────────────
import subprocess, sys, os, time, webbrowser, threading

app_path = os.path.join(os.getcwd(), 'app.py')
PORT = 8501

def open_browser():
    time.sleep(2.5)
    webbrowser.open(f'http://localhost:{PORT}')

print('🚀 Starting AI-Powered Invoice Processing & Entity Extraction Dashboard...')
threading.Thread(target=open_browser, daemon=True).start()

subprocess.run([
    sys.executable, '-m', 'streamlit', 'run', app_path,
    '--server.port', str(PORT),
    '--server.headless', 'true',
    '--browser.gatherUsageStats', 'false',
    '--theme.base', 'light',
])
```
